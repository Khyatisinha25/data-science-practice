# train test split and model training

```python
#train test split 

X=shop_data.drop(["Revenue","Browser","TrafficType","Region","OperatingSystems"],axis=1)
y=shop_data["Revenue"]
```

```python
X_train,X_test,y_train,y_test=train_test_split(
    X,y,test_size=0.3,random_state=42
)
```

## Model Training
### Baseline Decision Tree (No Pruning) → then CCP Pruning to improve F1 Score

```python
#Decision tree model-no pruning

from sklearn.tree import DecisionTreeClassifier
model=DecisionTreeClassifier()
model.fit(X_train,y_train)

y_pred=model.predict(X_test)

from sklearn.metrics import accuracy_score,f1_score
print("accuracy :",accuracy_score(y_test,y_pred))
print("f1 score :",f1_score(y_test,y_pred))
```

```python
#Decision tree with pre pruning

full_tree = DecisionTreeClassifier(random_state=42)
full_tree.fit(X_train, y_train)
```

```python
path = full_tree.cost_complexity_pruning_path(X_train, y_train)
ccp_alphas = path.ccp_alphas

print(ccp_alphas)
```

```python
#train our model for all alphas

trees = []

for alpha in ccp_alphas:
    model = DecisionTreeClassifier(random_state=42, ccp_alpha=alpha,class_weight='balanced')
    model.fit(X_train, y_train)

    trees.append((model, alpha))
```

```python
best_acc = 0
best_alpha = 0
best_f1 = 0

for model, alpha in trees:
    curr_acc = model.score(X_test, y_test)
    y_pred = model.predict(X_test)
    curr_f1 = f1_score(y_test, y_pred)
    
    if curr_f1 > best_f1:
        best_f1 = curr_f1
        best_alpha = alpha
        best_acc = curr_acc

print(f"Best Alpha: {best_alpha}")
print(f"Best F1 Score: {best_f1}")
print(f"Accuracy at best F1: {best_acc}")
```

```python
best_model = DecisionTreeClassifier(ccp_alpha=best_alpha,class_weight='balanced')
best_model.fit(X_train, y_train)
```

```python
from sklearn.tree import plot_tree
plt.figure(figsize=(18, 10))
plot_tree(
    best_model,
    feature_names=X.columns,
    filled=True
)
        
plt.tight_layout()
plt.show()
 
```

```python
# predict with best model
y_final_pred = best_model.predict(X_test)


print(f"Accuracy: {accuracy_score(y_test, y_final_pred):.4f}")
print(f"F1 Score: {f1_score(y_test, y_final_pred):.4f}")
```

## Conclusion
- Baseline Decision Tree: high accuracy but F1 ~0.55 due to class imbalance
- CCP Pruning + class_weight='balanced' improved F1 score significantly
- PageValues identified as most important feature by the model
- Pruning reduced overfitting while maintaining prediction quality
